Paired boostrap helpers for Accuracy and F1

```python
def paired_bootstrap_accuracy(
    df,
    gold_col,
    theory_col,
    oper_col,
    n_boot=10_000,
    seed=45
):
    
    # --------------------------------------------------
    # REMOVE MISSING PREDICTIONS
    # --------------------------------------------------
    
    data = df[
        [gold_col, theory_col, oper_col]
    ].dropna().reset_index(drop=True)
    
    n = len(data)
    
    gold = data[gold_col].to_numpy()
    theory = data[theory_col].to_numpy()
    oper = data[oper_col].to_numpy()
    
    
    # --------------------------------------------------
    # ORIGINAL POINT ESTIMATES
    # --------------------------------------------------
    
    theory_acc = np.mean(theory == gold)
    oper_acc = np.mean(oper == gold)
    
    observed_diff = oper_acc - theory_acc
    
    
    # --------------------------------------------------
    # PAIRED BOOTSTRAP
    # --------------------------------------------------
    
    rng = np.random.default_rng(seed)
    
    bootstrap_diffs = np.empty(n_boot)
    
    for b in range(n_boot):
        
        # Sample ARTICLE INDICES with replacement
        idx = rng.choice(
            n,
            size=n,
            replace=True
        )
        
        # Same indices used for gold + BOTH prompts
        gold_b = gold[idx]
        theory_b = theory[idx]
        oper_b = oper[idx]
        
        theory_acc_b = np.mean(
            theory_b == gold_b
        )
        
        oper_acc_b = np.mean(
            oper_b == gold_b
        )
        
        bootstrap_diffs[b] = (
            oper_acc_b - theory_acc_b
        )
    
    
    # --------------------------------------------------
    # 95% PERCENTILE BOOTSTRAP CI
    # --------------------------------------------------
    
    ci_low, ci_high = np.percentile(
        bootstrap_diffs,
        [2.5, 97.5]
    )
    
    
    # --------------------------------------------------
    # RETURN RESULTS
    # --------------------------------------------------
    
    return {
        "n": n,
        "theoretical_accuracy": theory_acc,
        "operationalized_accuracy": oper_acc,
        "observed_difference": observed_diff,
        "ci_95_low": ci_low,
        "ci_95_high": ci_high,
        "bootstrap_differences": bootstrap_diffs,
    }

from sklearn.metrics import f1_score
import numpy as np

def paired_bootstrap_f1(
    df,
    gold_col,
    theory_col,
    oper_col,
    average="macro",
    n_boot=10_000,
    seed=45
):
    
    data = df[
        [gold_col, theory_col, oper_col]
    ].dropna().reset_index(drop=True)
    
    n = len(data)
    
    gold = data[gold_col].to_numpy()
    theory = data[theory_col].to_numpy()
    oper = data[oper_col].to_numpy()

    # Use the classes represented in the original gold sample
    labels = np.unique(gold)
    
    
    # --------------------------------------------------
    # ORIGINAL-SAMPLE F1
    # --------------------------------------------------
    
    theory_f1 = f1_score(
        gold,
        theory,
        labels=labels,
        average=average,
        zero_division=0
    )
    
    oper_f1 = f1_score(
        gold,
        oper,
        labels=labels,
        average=average,
        zero_division=0
    )
    
    observed_diff = oper_f1 - theory_f1
    
    
    # --------------------------------------------------
    # PAIRED BOOTSTRAP
    # --------------------------------------------------
    
    rng = np.random.default_rng(seed)
    
    bootstrap_diffs = np.empty(n_boot)
    
    for b in range(n_boot):
        
        idx = rng.choice(
            n,
            size=n,
            replace=True
        )
        
        gold_b = gold[idx]
        theory_b = theory[idx]
        oper_b = oper[idx]
        
        
        theory_f1_b = f1_score(
            gold_b,
            theory_b,
            labels=labels,
            average=average,
            zero_division=0
        )
        
        oper_f1_b = f1_score(
            gold_b,
            oper_b,
            labels=labels,
            average=average,
            zero_division=0
        )
        
        bootstrap_diffs[b] = oper_f1_b - theory_f1_b
    
    
    # --------------------------------------------------
    # 95% PERCENTILE CI
    # --------------------------------------------------
    
    ci_low, ci_high = np.percentile(
        bootstrap_diffs,
        [2.5, 97.5]
    )
    
    
    return {
        "n": n,
        "theoretical_f1": theory_f1,
        "operationalized_f1": oper_f1,
        "f1_difference": observed_diff,
        "bootstrap_ci_low": ci_low,
        "bootstrap_ci_high": ci_high,
        "bootstrap_differences": bootstrap_diffs,
    }
```

# Four Binary Tasks with GPT-5 Mini: Using Paired Bootstrap and McNemar's (Holm correction)

Load and align

```python
import pandas as pd
import numpy as np

# --------------------------------------------------
# LOAD FULL 120-ROW RESULTS
# --------------------------------------------------


theory = pd.read_excel(
    "gpt5-mini_scores/theoretical_binary_joint_gpt-5-mini.xlsx"
)

oper = pd.read_excel(
    "gpt5-mini_scores/operationalized_binary_joint_gpt-5-mini.xlsx"
)

print("Theoretical rows:", len(theory))
print("Operationalized rows:", len(oper))


# --------------------------------------------------
# TASKS
# --------------------------------------------------

tasks = [
    "demands_present",
    "protester_quote",
    "authority_quote",
    "thematic_present",
]


# --------------------------------------------------
# KEEP NEEDED COLUMNS
# --------------------------------------------------

theory = theory[
    ["uuid"] +
    tasks +
    [f"gpt_{task}" for task in tasks]
].copy()

oper = oper[
    ["uuid"] +
    [f"gpt_{task}" for task in tasks]
].copy()


# --------------------------------------------------
# RENAME PREDICTION COLUMNS
# --------------------------------------------------

theory = theory.rename(columns={
    f"gpt_{task}": f"theory_{task}"
    for task in tasks
})

oper = oper.rename(columns={
    f"gpt_{task}": f"oper_{task}"
    for task in tasks
})


# --------------------------------------------------
# MERGE BY UUID
# --------------------------------------------------

compare_df = theory.merge(
    oper,
    on="uuid",
    how="inner",
    validate="one_to_one"
)

print("Paired articles:", len(compare_df))

display(compare_df.head())
```

Bootstrap helper

```python
def paired_bootstrap_accuracy(
    df,
    gold_col,
    theory_col,
    oper_col,
    n_boot=10_000,
    seed=45
):
    
    data = df[
        [gold_col, theory_col, oper_col]
    ].dropna().reset_index(drop=True)
    
    n = len(data)
    
    gold = data[gold_col].to_numpy()
    theory = data[theory_col].to_numpy()
    oper = data[oper_col].to_numpy()
    
    
    # --------------------------------------------------
    # ORIGINAL SAMPLE POINT ESTIMATES
    # --------------------------------------------------
    
    theory_acc = np.mean(
        theory == gold
    )
    
    oper_acc = np.mean(
        oper == gold
    )
    
    observed_diff = (
        oper_acc - theory_acc
    )
    
    
    # --------------------------------------------------
    # PAIRED BOOTSTRAP
    # --------------------------------------------------
    
    rng = np.random.default_rng(seed)
    
    bootstrap_diffs = np.empty(
        n_boot
    )
    
    for b in range(n_boot):
        
        idx = rng.choice(
            n,
            size=n,
            replace=True
        )
        
        gold_b = gold[idx]
        theory_b = theory[idx]
        oper_b = oper[idx]
        
        theory_acc_b = np.mean(
            theory_b == gold_b
        )
        
        oper_acc_b = np.mean(
            oper_b == gold_b
        )
        
        bootstrap_diffs[b] = (
            oper_acc_b
            - theory_acc_b
        )
    
    
    # --------------------------------------------------
    # 95% PERCENTILE CI
    # --------------------------------------------------
    
    ci_low, ci_high = np.percentile(
        bootstrap_diffs,
        [2.5, 97.5]
    )
    
    
    return {
        "n": n,
        "theoretical_accuracy": theory_acc,
        "operationalized_accuracy": oper_acc,
        "accuracy_difference": observed_diff,
        "bootstrap_ci_low": ci_low,
        "bootstrap_ci_high": ci_high,
        "bootstrap_differences": bootstrap_diffs,
    }
```

Exact McNemar's

```python
from statsmodels.stats.contingency_tables import mcnemar

def run_mcnemar(
    df,
    gold_col,
    theory_col,
    oper_col
):
    
    data = df[
        [gold_col, theory_col, oper_col]
    ].dropna()
    
    gold = data[gold_col]
    
    theory_correct = (
        data[theory_col] == gold
    )
    
    oper_correct = (
        data[oper_col] == gold
    )
    
    
    # --------------------------------------------------
    # 2x2 CORRECTNESS TABLE
    #
    #                    Operationalized
    #                 Wrong       Correct
    # Theory Wrong      a            b
    # Theory Correct    c            d
    # --------------------------------------------------
    
    both_wrong = (
        (~theory_correct) &
        (~oper_correct)
    ).sum()
    
    oper_only_correct = (
        (~theory_correct) &
        oper_correct
    ).sum()
    
    theory_only_correct = (
        theory_correct &
        (~oper_correct)
    ).sum()
    
    both_correct = (
        theory_correct &
        oper_correct
    ).sum()
    
    
    table = [
        [
            both_wrong,
            oper_only_correct
        ],
        [
            theory_only_correct,
            both_correct
        ]
    ]
    
    
    # Exact McNemar is appropriate for small discordant counts
    result = mcnemar(
        table,
        exact=True
    )
    
    
    return {
        "both_wrong": both_wrong,
        "oper_only_correct": oper_only_correct,
        "theory_only_correct": theory_only_correct,
        "both_correct": both_correct,
        "mcnemar_statistic": result.statistic,
        "mcnemar_p_raw": result.pvalue,
    }
```

Run all four tasks and Holm correction

```python
from statsmodels.stats.multitest import multipletests

results = []

bootstrap_objects = {}


for task in tasks:
    
    print(f"Running: {task}")
    
    
    # --------------------------------------------------
    # PAIRED BOOTSTRAP
    # --------------------------------------------------
    
    boot = paired_bootstrap_accuracy(
        df=compare_df,
        gold_col=task,
        theory_col=f"theory_{task}",
        oper_col=f"oper_{task}",
        n_boot=10_000,
        seed=45
    )
    
    bootstrap_objects[task] = boot
    
    
    # --------------------------------------------------
    # MCNEMAR
    # --------------------------------------------------
    
    mc = run_mcnemar(
        df=compare_df,
        gold_col=task,
        theory_col=f"theory_{task}",
        oper_col=f"oper_{task}"
    )
    
    
    results.append({
        "task": task,
        
        "n": boot["n"],
        
        "theoretical_accuracy":
            boot["theoretical_accuracy"],
        
        "operationalized_accuracy":
            boot["operationalized_accuracy"],
        
        "difference_oper_minus_theory":
            boot["accuracy_difference"],
        
        "bootstrap_ci_95_low":
            boot["bootstrap_ci_low"],
        
        "bootstrap_ci_95_high":
            boot["bootstrap_ci_high"],
        
        "theory_only_correct":
            mc["theory_only_correct"],
        
        "oper_only_correct":
            mc["oper_only_correct"],
        
        "mcnemar_p_raw":
            mc["mcnemar_p_raw"],
    })


summary_df = pd.DataFrame(
    results
)


# --------------------------------------------------
# HOLM MULTIPLE-TEST CORRECTION
# --------------------------------------------------

reject, pvals_holm, _, _ = multipletests(
    summary_df["mcnemar_p_raw"],
    alpha=0.05,
    method="holm"
)

summary_df["mcnemar_p_holm"] = (
    pvals_holm
)

summary_df["mcnemar_reject_holm"] = (
    reject
)


# --------------------------------------------------
# ROUND FOR DISPLAY
# --------------------------------------------------

display_cols = [
    "task",
    "n",
    "theoretical_accuracy",
    "operationalized_accuracy",
    "difference_oper_minus_theory",
    "bootstrap_ci_95_low",
    "bootstrap_ci_95_high",
    "theory_only_correct",
    "oper_only_correct",
    "mcnemar_p_raw",
    "mcnemar_p_holm",
    "mcnemar_reject_holm",
]

display(
    summary_df[
        display_cols
    ].round(3)
)
```

Helper for confusion matrix

```python
def plot_frame_confusion_matrix(
    df,
    gold_col,
    pred_col,
    title
):
    
    valid = df[
        [gold_col, pred_col]
    ].dropna()
    
    cm = confusion_matrix(
        valid[gold_col],
        valid[pred_col],
        labels=FRAME_LABELS
    )
    
    cm_df = pd.DataFrame(
        cm,
        index=FRAME_LABELS,
        columns=FRAME_LABELS
    )
    
    fig, ax = plt.subplots(
        figsize=(11, 9)
    )
    
    im = ax.imshow(cm)
    
    ax.set_xticks(
        range(len(FRAME_LABELS))
    )
    ax.set_yticks(
        range(len(FRAME_LABELS))
    )
    
    ax.set_xticklabels(
        FRAME_LABELS,
        rotation=45,
        ha="right"
    )
    
    ax.set_yticklabels(
        FRAME_LABELS
    )
    
    ax.set_xlabel(
        "GPT-5 Prediction"
    )
    
    ax.set_ylabel(
        "Human Gold Label"
    )
    
    ax.set_title(title)
    
    
    # Add counts inside cells
    for i in range(len(FRAME_LABELS)):
        for j in range(len(FRAME_LABELS)):
            
            value = cm[i, j]
            
            ax.text(
                j,
                i,
                str(value),
                ha="center",
                va="center"
            )
    
    fig.colorbar(
        im,
        ax=ax,
        label="Count"
    )
    
    plt.tight_layout()
    plt.show()
    
    return cm_df
```

Frequency of each title and excerpt frame, how accurate GPT was, and recall

```python
def frame_recall_summary(
    df,
    gold_col,
    pred_col
):
    
    rows = []
    
    for frame in FRAME_LABELS:
        
        subset = df[
            df[gold_col] == frame
        ]
        
        n_gold = len(subset)
        
        n_correct = (
            subset[pred_col] == frame
        ).sum()
        
        n_missed = (
            n_gold - n_correct
        )
        
        recall = (
            n_correct / n_gold
            if n_gold > 0
            else np.nan
        )
        
        rows.append({
            "frame": frame,
            "gold_n": n_gold,
            "correct": n_correct,
            "missed": n_missed,
            "recall": recall,
        })
    
    return pd.DataFrame(rows)
```
