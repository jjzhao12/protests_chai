```python
import pandas as pd
import numpy as np

frame_col = "gpt_excerpt_primary_frame"
lean_col = "political_leaning"

lean_order = ["left", "left_lean", "center", "right_lean", "right"]

# Primary frame distribution
frame_dist = pd.DataFrame({
    "N": df[frame_col].value_counts(),
    "Percent": df[frame_col].value_counts(normalize=True) * 100
}).round(1)

# Political leaning distribution
lean_dist = pd.DataFrame({
    "N": df[lean_col].value_counts(),
    "Percent": df[lean_col].value_counts(normalize=True) * 100
}).reindex(lean_order).round(1)

print("PRIMARY FRAME DISTRIBUTION")
display(frame_dist)

print("POLITICAL LEANING DISTRIBUTION")
display(lean_dist)
```

```python
frame_counts = pd.crosstab(
    df[lean_col],
    df[frame_col]
).reindex(lean_order, fill_value=0)

frame_pct = pd.crosstab(
    df[lean_col],
    df[frame_col],
    normalize="index"
).reindex(lean_order).mul(100).round(1)

print("COUNTS")
display(frame_counts)

print("ROW PERCENTAGES — Within each political leaning")
display(frame_pct)
```

Generally, right leaning and right outlets opt more for primary frame of disruption, property damage and violence, while left lean opts for grievance and harm to protesters. 

```python
binary_cols = [
    "gpt_demands_present",
    "gpt_protester_quote",
    "gpt_authority_quote",
    "gpt_thematic_present"
]

binary_rates = (
    df.groupby(lean_col)[binary_cols]
    .mean()
    .reindex(lean_order)
    .mul(100)
    .round(1)
)

binary_counts = (
    df.groupby(lean_col)[binary_cols]
    .sum()
    .reindex(lean_order)
)

print("FEATURE PREVALENCE (%)")
display(binary_rates)

print("FEATURE COUNTS (Number with feature = 1)")
display(binary_counts)
```

```python
from scipy.stats import chi2_contingency
from statsmodels.stats.multitest import multipletests

results = []

for feature in binary_cols:

    temp = df[[lean_col, feature]].dropna()

    ct = pd.crosstab(
        temp[lean_col],
        temp[feature]
    ).reindex(
        index=lean_order,
        columns=[0, 1],
        fill_value=0
    )

    # Remove empty leaning categories
    ct = ct.loc[ct.sum(axis=1) > 0]

    if len(ct) < 2 or (ct.sum(axis=0) == 0).any():
        continue

    chi2, p, dof, expected = chi2_contingency(ct)

    results.append({
        "Feature": feature,
        "N": len(temp),
        "Chi-square": chi2,
        "df": dof,
        "p-value": p,
        "Min expected": expected.min()
    })

binary_tests = pd.DataFrame(results)

binary_tests["p-adjusted"] = multipletests(
    binary_tests["p-value"],
    method="fdr_bh"
)[1]

display(binary_tests.round(4))
```

# Distinguishing between pro-israel and pro-palestine protest articles

```python
import pandas as pd

df = israel_hamas_df.copy()

text = (
    df["title"].fillna("").astype(str) + " " +
    df["protest_excerpt"].fillna("").astype(str)
)

pal_pattern = (
    r"\bpro[\s-]?palestin(?:ian|e)\b"
    r"|\banti[\s-]?israel(?:i)?\b"
    r"|\bfree palestine\b"
    r"|\bpalestinian solidarity\b"
    r"|\bpro[\s-]?gaza\b"
    r"|\bgaza solidarity\b"
    r"|\bceasefire (?:protest|rally|demonstration)\b"
    r"|\b(?:protest|rally|demonstration)\w*\s+(?:for|demanding|calling for)\s+(?:a\s+)?ceasefire\b"
)

israel_pattern = (
    r"\bpro[\s-]?israel(?:i)?\b"
    r"|\bstand with israel\b"
    r"|\bsupport(?:ers|ing)? israel\b"
    r"|\bisrael solidarity\b"
    r"|\bbring (?:them|the hostages) home\b"
    r"|\b(?:protest|rally|demonstration)\w*\s+(?:for|demanding|calling for)\s+(?:the\s+)?release of (?:the\s+)?hostages\b"
)

df["pal_match"] = text.str.contains(
    pal_pattern, case=False, regex=True, na=False
)

df["israel_match"] = text.str.contains(
    israel_pattern, case=False, regex=True, na=False
)

df["regex_group"] = "neither"
df.loc[df["pal_match"], "regex_group"] = "pal_only"
df.loc[df["israel_match"], "regex_group"] = "israel_only"
df.loc[
    df["pal_match"] & df["israel_match"],
    "regex_group"
] = "both"

display(df["regex_group"].value_counts())

lean_order = ["left", "left_lean", "center", "right_lean", "right"]

display(
    pd.crosstab(
        df["political_leaning"],
        df["regex_group"],
        normalize="index"
    )
    .reindex(lean_order)
    .mul(100)
    .round(1)
)
```

Subset to only protests where Pro-Palestine or Anti-Israel is mentioned

```python
pal_df = df[df["regex_group"] == "pal_only"].copy()

lean_order = ["left", "left_lean", "center", "right_lean", "right"]

print(f"Pal-only articles: {len(pal_df):,}")

# Sample sizes by political leaning
display(
    pal_df["political_leaning"]
    .value_counts()
    .reindex(lean_order, fill_value=0)
)

# Primary frame distribution by leaning
pal_frame_pct = pd.crosstab(
    pal_df["political_leaning"],
    pal_df["gpt_excerpt_primary_frame"],
    normalize="index"
).reindex(lean_order).mul(100).round(1)

display(pal_frame_pct)
```

```python
delegit_frames = [
    "protester_violence",
    "property_damage",
    "disruption"
]

pal_df["delegitimizing"] = (
    pal_df["gpt_excerpt_primary_frame"]
    .isin(delegit_frames)
    .astype(int)
)

delegit_summary = (
    pal_df.groupby("political_leaning")["delegitimizing"]
    .agg(
        N="size",
        delegit_N="sum",
        delegit_rate="mean"
    )
    .reindex(lean_order)
)

delegit_summary["delegit_rate"] *= 100

display(delegit_summary.round(1))
```

```python
import statsmodels.formula.api as smf
import numpy as np

lean_map = {
    "left": -2,
    "left_lean": -1,
    "center": 0,
    "right_lean": 1,
    "right": 2
}

pal_df["lean_score"] = (
    pal_df["political_leaning"].map(lean_map)
)

model = smf.logit(
    "delegitimizing ~ lean_score",
    data=pal_df
).fit(disp=False)

ci = model.conf_int().loc["lean_score"]

print("OR per step rightward:", round(np.exp(model.params["lean_score"]), 3))
print("95% CI:", np.exp(ci).round(3).tolist())
print("p-value:", round(model.pvalues["lean_score"], 4))
```
